```python
import os
import json
import pandas as pd
from datetime import datetime
import ipywidgets as widgets
from IPython.display import display, JSON, clear_output

# Main function to extract column and convert to JSON
def extract_column_to_json(csv_path, column_name=None, column_index=None, output_format='array', include_header=True):
  
    result = {
        "metadata": { # replace with whatever metadata we want
            "source_file": os.path.basename(csv_path),
            "conversion_date": datetime.now().isoformat(),
            "format": output_format
        }
    }
    
    # Read the CSV file to dataframe
    df = pd.read_csv(csv_path)
    
    # Get the column data
    if column_name is not None:
        if column_name in df.columns:
            column_data = df[column_name]
            result["metadata"]["column_name"] = column_name
        else:
            raise ValueError(f"Column '{column_name}' not found in CSV file")

    # Format the data according to the specified output format
    if output_format == 'array':
        # Simple array of values
        result["data"] = column_data.tolist()
        
    elif output_format == 'objects':
        # Array of objects with value property
        result["data"] = [{"value": val} for val in column_data.tolist()]
        
    elif output_format == 'key_value':
        # Object with row indices as keys and column values as values
        result["data"] = {str(i): val for i, val in enumerate(column_data.tolist())}
        
    elif output_format == 'numbered':
        # Array of objects with index and value properties
        result["data"] = [{"index": i, "value": val} for i, val in enumerate(column_data.tolist())]
        
    else:
        # Default to simple array
        result["data"] = column_data.tolist()
    
    # Include header information if requested
    if include_header:
        result["header"] = column_name
    
    return result

# Function to save JSON to file
def save_json_to_file(json_data, output_path):
    """Save JSON data to a file"""
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(json_data, f, indent=2, ensure_ascii=False)
    print(f"Saved to: {output_path}")

# Function to list columns in CSV
def list_columns(csv_path):
    """List all columns in a CSV file with their indices"""
    df = pd.read_csv(csv_path)
    columns_info = [(i, col) for i, col in enumerate(df.columns)]
    return columns_info
```
