```python
from pathlib import Path

# Path to your main folder
base_path = Path("copied folders")

# Create a set of valid relative paths from your DataFrame
allowed_paths = set(df.apply(lambda row: f"{row.folder_name}/{row.file_name}", axis=1))

# Find all PDFs recursively
all_pdfs = [p for p in base_path.rglob("*.pdf")]

# Identify PDFs not in your DataFrame
extra_pdfs = [p for p in all_pdfs if str(p.relative_to(base_path)) not in allowed_paths]

print(f"Found {len(extra_pdfs)} extra PDFs. Deleting...")

# Delete the extra PDFs
for pdf in extra_pdfs:
    try:
        pdf.unlink()
        print(f"Deleted: {pdf}")
    except Exception as e:
        print(f"Could not delete {pdf}: {e}")

print("Cleanup complete ✅")
```
