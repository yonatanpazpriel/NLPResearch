```python
import pandas as pd
```

```python
df = pd.read_csv('all_files.csv')
```

```python
df = df.drop(["year_month", "month"], axis=1)
df
```

```python
df[df.isna().any(axis=1)]
```

```python
manually_dated = pd.read_csv("manually_dated.csv") 

# Set multi-index on keys for both DataFrames
df.set_index(["file_name", "folder_name"], inplace=True)
manually_dated.set_index(["file_name", "folder_name"], inplace=True)

# Use .update() to replace rows in df_main with df_update values
df.update(manually_dated)

# Reset index back to columns
df.reset_index(inplace=True)

df
```

```python
df[df.isna().any(axis=1)]
```

```python
df = df.dropna()
df
```

```python
df["year_month"] = df["date"].str.slice(0, 7)  # "YYYY-MM"
df
```

```python
for period, group in df.groupby("year_month"):
    filename = str(period).replace("-", "_") + ".csv"  # "2019_01.csv"
    group.to_csv(filename, index=False)
```
